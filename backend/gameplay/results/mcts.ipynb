```python
import pandas as pd
import numpy as py
import matplotlib.pyplot as plt

df = pd.read_csv("mcts.csv")
df.head()

x = df["rollouts"]
wins = df["win_rate"]
times = df["time_taken"]
```

```python
plt.plot(x, wins)
plt.title("Effect of number of rollouts on win rate within MCTS")
plt.show()

plt.plot(x, times)
plt.title("Graph depicting how number of rollouts within MCTS affects time taken")
plt.show()
```
