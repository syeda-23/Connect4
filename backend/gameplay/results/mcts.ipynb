```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('mcts.csv')

df.head
print(df.iloc[0])
x = df["rollouts"]
wins = df["win_rate"]
times = df["time_taken"]

plt.plot(x, wins)
plt.title("Effect of number of rollouts on \nwin rate within MCTS")
plt.xlabel("Number of rollouts per move")
plt.ylabel("Win rate against baseline MCTS opponent with 500 rollouts")
plt.show()

plt.plot(x, times)
plt.title("Graph depicting how number of rollouts \nwithin MCTS affects time taken")
plt.xlabel("Number of rollouts per move")
plt.ylabel("Average time taken to make a move")
plt.show()
```
