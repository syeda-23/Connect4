```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('minimax.csv')

df.head
print(df.iloc[0])
x = df["depth"]
wins = df[" win_rate"]
nodes = df[" average_nodes"]
times = df[" average_time"]
```

```python
plt.scatter(x, nodes)
plt.title("Graph depicting how the number of nodes explored \nvaries across different minimax depths")
plt.xlabel("Depth")
plt.ylabel("Average number of nodes explored each move")
plt.show()

plt.scatter(x, wins)
plt.title("Graph depicting how varying minimax depth \naffects win rate in Connect 4")
plt.xlabel("Depth")
plt.ylabel("Average win rate per 100 games")
plt.show()

plt.scatter(x, times)
plt.title("Graph depicting average time taken to make \na move across varying minimax depths")
plt.xlabel("Depth")
plt.ylabel("Average times taken per move")
plt.show()

```
